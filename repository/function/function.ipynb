```python
def add(*args):
    s = 0
    for i in args:
        s = s+i
    return s
```

```python
add(23,45,67,78)
```

```python
sum_args = add(23,45,67,78)
print(sum_args)
```

```python
sum_a1 = add(25)
print(sum_a1)
```

```python
def multiply(x,*args):
    p=1
    for i in args:
        p = p*i
    return f"product performed by {x} is {p}"
```
